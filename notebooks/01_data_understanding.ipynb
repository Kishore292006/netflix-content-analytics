### Insight: Netflix Titles Added by Year

The number of titles added to the Netflix catalog increased substantially from 2015, reaching its highest level in 2019. The number of additions then declined in 2020 and 2021. Very few titles in the dataset were recorded as added between 2008 and 2014.

**Business observation:** The catalog additions were concentrated in the later years represented in the dataset. This can help describe the timing of catalog expansion, but it does not establish actual viewership or audience preferences.

## Final Summary: Data Understanding

### 1. Dataset Overview
The Netflix dataset contains 8,807 records and 12 columns describing movies and TV shows. It includes information about content type, title, director, cast, country, date added, release year, rating, duration, genres, and descriptions.

### 2. Key Observations
- Movies account for approximately 69.62% of the dataset, while TV shows account for 30.38%.
- The number of titles by release year is concentrated in more recent years, with 2018 having the highest count in the dataset.
- Catalog additions peaked in 2019, followed by a decline in 2020 and 2021.
- TV-MA is the most common content rating, followed by TV-14.
- The United States has the highest number of associated titles, followed by India and the United Kingdom.
- One-season TV shows are more common than shows with multiple seasons.
- Dramas, International Movies is the most frequent exact genre/category combination.

### 3. Data-Quality Findings
- Missing values occur in director, cast, country, date_added, rating, and duration.
- No duplicate rows or duplicate show IDs were identified.
- The date_added column can be converted to a datetime data type, with 10 missing values remaining.
- Missing values and any inconsistent or invalid records must be investigated during the cleaning phase.

### 4. Limitations
This dataset describes Netflix catalog metadata. It does not provide actual viewing figures, subscriber counts, streaming revenue, or audience engagement. Therefore, catalog counts should not be interpreted as measures of popularity or viewer demand.
