```python
# Check whether durations match the expected format
movie_mask=df['type'].eq('Movie')
tv_mask=df['type'].eq('TV Show')
invalid_movies=df[
    movie_mask & ~df['duration'].str.fullmatch(r'\d+ min',na=False)
]
invalid_tv_shows=df[
    tv_mask & ~df['duration'].str.fullmatch(r'\d+ Seasons?',na=False)
]
print("Invalid movie durations:", len(invalid_movies))
print("Invalid TV Show durations:", len(invalid_tv_shows))
```

```python
print('Unxpected content types:')
print(df.loc[~df['type'].isin(['Movie', 'TV Show']), 'type'].unique())

#check release years
print('\nInvalid release years:')
print(df.loc[(df['release_year']<1925) | (df['release_year']>2021),['title','release_year']])
```

## Data Cleaning Summary

### Objective
The objective of this stage was to improve the quality and consistency of the Netflix dataset before performing exploratory data analysis (EDA).

### Cleaning Steps Performed
- **Missing Values:** Replaced missing values in `director`, `cast`, `country`, and `rating` with `"Unknown"`. Preserved the 10 missing `date_added` values as missing because their original dates could not be determined.
- **Data Correction:** Identified three movie records where duration values were incorrectly stored in the `rating` column and moved those values to `duration`.
- **Duplicate Checks:** Verified that there were no duplicate rows or duplicate `show_id` values.
- **Data Validation:** Checked content types, release years, rating categories, and duration formats for inconsistencies.
- **Date Conversion:** Converted `date_added` into a datetime column to support time-based analysis.

### Final Dataset Status
- Total records: 8,807
- Total columns: 12
- Duplicate rows: 0
- Missing values: 10, all in `date_added`
- Movie and TV Show duration formats: Passed validation

### Limitations
The missing values replaced with `"Unknown"` remain unknown; the original information could not be recovered. The three corrected duration values were reassigned based on the inspected records. The validation checks confirm consistency with the rules used, but do not guarantee that every source value is factually correct.

### Conclusion
The dataset is now prepared for exploratory data analysis. The original raw dataset was retained separately, and the cleaned dataset was saved as `data/processed/netflix_cleaned.csv`.
